## STEP 3. 개인별 AI 활용과 구현·수정

- 이름별 구성: 목적·완료 기준 → AI 요청·반영 판단 → 기존 코드·시험 입력 → 수정·재실행 → 개인별 참여 기록
- 프롬프트 표현: 담당 관점에 맞춰 말투와 길이를 다듬은 요청문
- 기록 범위: 실제 학생별 대화 이력을 나타내는 자료는 아님


```python
from pathlib import Path
import ast
import re
import subprocess
import tempfile
from types import SimpleNamespace

BASE = "cf389690fd36f457d7ddeab4ae2b323029d348aa"
WORK = Path(tempfile.mkdtemp(prefix="minime-simple-"))
REPO = WORK / "source"
subprocess.run(["git", "clone", "--quiet", "https://github.com/jkzk1397/minime.git", str(REPO)], check=True)
subprocess.run(["git", "checkout", "--quiet", "--detach", BASE], cwd=REPO, check=True)
FILES = ["app/persona.py", "app/sources.py", "app/ko.py"]
BASE_TEXT = {path: (REPO / path).read_text(encoding="utf-8") for path in FILES}
CHANGES = {}

def function_source(text, name):
    node = next(n for n in ast.parse(text).body
                if isinstance(n, ast.FunctionDef) and n.name == name)
    return "\n".join(text.splitlines()[node.lineno - 1:node.end_lineno])

def load_functions(texts):
    # 실제 소스에서 네 함수만 추출해 실행한다. 웹 서버·AI 연결은 필요 없다.
    scope = {"re": re, "MAX_CHAT_LINES": 400}
    exec(function_source(texts["app/ko.py"], "sentences"), scope)
    scope["ko"] = SimpleNamespace(sentences=scope["sentences"])
    for path, name in [("app/persona.py", "split_report"),
                       ("app/persona.py", "split_pages"), ("app/sources.py", "clean_lines")]:
        exec(function_source(texts[path], name), scope)
    return scope

def changed(texts, change):
    path, old, new = change
    result = dict(texts)
    if result[path].count(old) != 1:
        raise ValueError("수정 대상 코드가 정확히 한 곳이어야 합니다.")
    result[path] = result[path].replace(old, new, 1)
    return result

RESULTS = []
def check(label, expected, actual):
    ok = expected == actual
    RESULTS.append(ok)
    print(label)
    # 긴 원문 대신 길이와 앞부분을 출력하고 판정은 전체 값으로 한다.
    def display(value):
        return f"문자열 {len(value)}자: {value[:30]}…" if isinstance(value, str) and len(value) > 100 else value
    print("  예상:", display(expected))
    print("  실제:", display(actual))
    print("  판정:", "통과" if ok else "개선 필요")

print("기준 코드 준비 완료. 파일 업로드·AI API 키·라이브러리 설치 없이 시험합니다.")
```

## 김동준 · 줄바꿈 형식 통일

**담당 작업·완료 기준**

- LF·CRLF·CR 입력에서 같은 두 문단을 반환한다.

**코드 위치**

- `app/persona.py`의 `split_report`. 아래 기존 코드와 수정 코드에 실제 변경 부분이 표시된다.

**AI 요청 · 담당별 요청문**

> MINIME 문서 입력에서 줄바꿈 때문에 결과가 달라지는 걸 고치고 싶어.
> 대상은 app/persona.py의 Python split_report이고, 다음 셀에서 출력하는 기존 함수 원문을 봐줘. 지금은 CRLF만 LF로 바꿔서 단독 CR로 붙인 두 문장이 따로 나뉘지 않아.
> 시험용 입력은 '첫 번째 시험용 문단입니다.'와 '두 번째 시험용 문단입니다.'를 LF·CRLF·CR로 이어 붙인 세 가지야. Colab에서 셋 다 같은 두 문단이 나오게 해 줘. 문단 내용과 순서는 그대로 남겨줘.
> 김혜중이 같은 함수의 긴 글 분할 구간을 고치니까 나는 줄바꿈 정규화 구간만 바꾸고 싶어. 송준서의 split_pages에서 계속 호출할 수 있게 함수 이름과 문자열 목록 반환 방식도 유지해 줘.

**입력·환경·연결 조건**

- 언어·도구: Python, Google Colab
- 제공 코드: 다음 코드 셀에 포함된 실제 기존 함수 원문
- 연결 조건: 기존 프로젝트의 함수 이름·호출 방식 유지

**AI 응답·채택 이유**

- CRLF를 LF로 바꾼 뒤 남은 CR을 LF로 바꾸면 두 글자 줄바꿈이 중복 변환되지 않는다.
- 기존 문단 분리 단계 앞에 적용한다.

**처리·수정 기준**

- CRLF를 먼저 LF로 바꾸고 남은 CR을 LF로 바꾼다.
- 문단 순서와 내용은 유지한다.


```python
# 이 셀의 입력 문장·길이를 바꾸어 추가 시험할 수 있다.
print(function_source(BASE_TEXT['app/persona.py'], 'split_report'))
TEST_1 = 'first = "첫 번째 시험용 문단입니다."\nsecond = "두 번째 시험용 문단입니다."\ninputs = [first + sep + second for sep in ["\\n", "\\r\\n", "\\r"]]\nfor label, text in zip(["LF", "CRLF", "CR"], inputs):\n    check(label, [first, second], split_report(text))\n'
before_scope_1 = load_functions(BASE_TEXT)
RESULTS.clear()
exec(TEST_1, dict(before_scope_1, check=check))
print("수정 전 통과:", sum(RESULTS), "/", len(RESULTS))
```

```python
OLD_1 = 'text = text.replace("\\r\\n", "\\n").strip()'
NEW_1 = 'text = text.replace("\\r\\n", "\\n").replace("\\r", "\\n").strip()'
CHANGES[1] = ('app/persona.py', OLD_1, NEW_1)
PERSONAL_1 = changed(BASE_TEXT, CHANGES[1])
print("수정 함수 원문:")
print(function_source(PERSONAL_1['app/persona.py'], 'split_report'))
after_scope_1 = load_functions(PERSONAL_1)
RESULTS.clear()
exec(TEST_1, dict(after_scope_1, check=check))
assert all(RESULTS)
print("수정 후 통과:", sum(RESULTS), "/", len(RESULTS))
```

- Colab에서 번호 순서대로 코드 셀을 실행한다. 입력 예시를 직접 바꿔 다시 실행할 수 있다. 변경 시 후속 개인 시험·교차 검증·최종 통합도 다시 실행한다.
- 전체 웹앱은 이 저장소를 내려받아 Windows run.bat 또는 macOS·Linux run.sh로 실행한다. 이번 개선의 동작 근거는 Colab의 실제 함수 출력이다.
- 외부 AI API 키·유료 서비스·시험 파일 업로드가 필요하지 않다. 최초 저장소 다운로드에 인터넷 연결이 필요하다.
- 실행 출력이 남은 Colab을 저장하고 평가자에게 열람 권한을 부여한다. **최종 제출물은 Colab 공유 링크 1개다.**
